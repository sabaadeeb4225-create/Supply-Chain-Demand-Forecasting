Step11 :- Creating the correlation matrix

```python
correlation = df[
    [
        "Inventory Level",
        "Units Sold",
        "Units Ordered",
        "Demand Forecast",
        "Price",
        "Revenue"
    ]
].corr()

correlation
```

Step 12 :- Creating the heatmap

```python
plt.figure(figsize=(9,7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.savefig("../images/correlation_heatmap.png", bbox_inches="tight")
plt.show()
```

A correlation heatmap is used to visualize the strength and direction of relationships between numerical variables. The colour indicates the strength of the correlation, while the number inside each cell gives the exact correlation coefficient. Values closer to +1 indicate a strong positive relationship, values around zero indicate little or no linear relationship, and negative values indicate an inverse relationship."

In our analysis, Units Sold and Demand Forecast have the strongest relationship, with a correlation of approximately 0.997, while Units Sold and Units Ordered have almost zero correlation."
