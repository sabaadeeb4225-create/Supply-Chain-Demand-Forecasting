Step 2 :- Monthly inventory trend

```python
monthly_inventory = df.groupby("Month_name")["Inventory Level"].sum()

monthly_inventory
```

Insight: January has the highest total inventory level (1,723,528), while February has the lowest (1,538,685). Inventory levels remain relatively stable throughout the year with only minor monthly fluctuations.

```python
Step 3 :- Creating a line chart 
```

```python
plt.figure(figsize=(10,5))

monthly_inventory.plot(
    kind="line",
    marker="o",
    color="blue"
)

plt.title("Monthly Inventory Trend")
plt.xlabel("Month")
plt.ylabel("Total Inventory Level")
plt.grid(True)

plt.show()
```

```python
Step 4 :- fixing the order of months in the line chart
```

```python
month_order = [
    "January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December"
]

monthly_inventory = monthly_inventory.reindex(month_order)

monthly_inventory
```

Week 3 :- Monthly units sold trend 
  Step 1:- Calculating Monthly Sales

```python
monthly_sales = df.groupby("Month_name")["Units Sold"].sum()
monthly_sales = monthly_sales.reindex(month_order)

monthly_sales
```

creating Month Order List 

```python
month_order = [
    "January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December"
]
```

Step 2 :- Creating the line chart for monthly unit sold trend

```python
plt.figure(figsize=(10,5))

monthly_sales.plot(
    kind="line",
    marker="o",
    color="green"
)

plt.title("Monthly Units Sold Trend")
plt.xlabel("Month")
plt.ylabel("Total Units Sold")

plt.grid(True)

plt.show()
```

Insight: The monthly sales trend helps identify periods of high and low customer demand. This information can help businesses plan inventory levels and promotional activities.

Step 3 :- Creating Revenue Column

```python
df["Revenue"] = df["Units Sold"] * df["Price"]
df.head()
```

Step 4 : Total Revenue by Category

```python
revenue_category = df.groupby("Category")["Revenue"].sum()

revenue_category
```

"Furniture generated the highest revenue at approximately 111.5 million, while Electronics generated the lowest revenue at approximately 108.4 million. The revenue values are quite close, indicating that the business earns revenue consistently across all product categories."

step 5 :- Plotting Revenue By Category 

```python
plt.figure(figsize=(8,5))

revenue_category.plot(
    kind="bar",
    color="purple"
)

plt.title("Revenue by Category")
plt.xlabel("Category")
plt.ylabel("Total Revenue")
plt.ylim(107000000, 112500000)

plt.xticks(rotation=0)

plt.grid(axis="y")

plt.show()
```

```python
Step 6:-Units sold by category 
```

```python

category_sales = df.groupby("Category")["Units Sold"].sum()

category_sales
```

Creating the chart

```python
plt.figure(figsize=(8,5))

category_sales.plot(
    kind="bar",
    color="skyblue",
    edgecolor="black"
)

plt.title("Total Units Sold by Category")
plt.xlabel("Category")
plt.ylabel("Units Sold")
plt.ylim(1900000, 2050000)

plt.xticks(rotation=0)
plt.grid(axis="y")

plt.show()
```

The chart compares total units sold across different product categories.

Furniture has the highest number of units sold, indicating relatively strong demand for this category. Electronics has the lowest units sold, while Clothing, Groceries, and Toys show similar sales volumes.

Overall, the differences in units sold across categories are relatively small, suggesting that customer demand is fairly balanced across the product categories. Inventory levels and replenishment decisions should therefore be monitored based on category-level demand.

Step 7 :-  Creating Chart for Inventory level by category 

```python
inventory_category = df.groupby("Category")["Inventory Level"].sum()

inventory_category
```

```python
plt.figure(figsize=(8,5))

inventory_category.plot(
    kind="bar",
    edgecolor="black"
)

plt.title("Total Inventory Level by Category")
plt.xlabel("Category")
plt.ylabel("Total Inventory Level")
plt.ylim(3900000, 4100000)


plt.xticks(rotation=0)
plt.grid(axis="y")

plt.show()
```

Furniture has the highest total inventory level among the categories, while Electronics has the lowest. The inventory levels across categories are relatively close, indicating a fairly balanced inventory distribution. Furniture may require closer monitoring because it holds the highest amount of inventory.
